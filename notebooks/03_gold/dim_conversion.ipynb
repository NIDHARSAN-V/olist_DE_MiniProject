```sql
use catalog ecommerce_dev;
use schema gold;
```

# DIM TABLE CREATION

```sql
create table if not exists ecommerce_dev.gold.dim_customer
(
   customer_id string,
   customer_unique_id string,
   customer_zip_code_prefix int
);
```

```sql
delete from ecommerce_dev.gold.dim_customer;

insert into ecommerce_dev.gold.dim_customer(customer_id,customer_unique_id,customer_zip_code_prefix)
select customer_id ,  customer_unique_id , customer_zip_code_prefix from ecommerce_dev.silver.customers_enhanced;

select * from ecommerce_dev.gold.dim_customer;
```

```sql
create table if not exists ecommerce_dev.gold.dim_product
(
   product_id string,
   product_category_name string,
   product_name_lenght int,
   product_description_lenght int,
   product_photos_qty int,
   product_weight_g int,
   product_length_cm int,
   product_height_cm int,
   product_width_cm int,
   is_category_missing  boolean
);
```

```sql
delete from ecommerce_dev.gold.dim_product;

insert into ecommerce_dev.gold.dim_product(product_id,product_category_name,product_name_lenght,product_description_lenght,product_photos_qty,product_weight_g,product_length_cm,product_height_cm,product_width_cm,is_category_missing)
select product_id, product_category_name, product_name_lenght, product_description_lenght, product_photos_qty, product_weight_g, product_length_cm, product_height_cm, product_width_cm, is_category_missing
from ecommerce_dev.silver.products_enhanced;

select * from ecommerce_dev.gold.dim_product;
```

```sql
create table if not exists ecommerce_dev.gold.dim_seller
(
    seller_id string,
    seller_zip_code_prefix int,
    seller_city string,
    seller_state string
);
```

```sql
delete from ecommerce_dev.gold.dim_seller;

insert into ecommerce_dev.gold.dim_seller(seller_id,seller_zip_code_prefix,seller_city,seller_state)
select seller_id,seller_zip_code_prefix,seller_city,seller_state
from ecommerce_dev.silver.sellers_enhanced;

select * from ecommerce_dev.gold.dim_seller;
```

```sql
create table if not exists ecommerce_dev.gold.dim_date
(
   order_id string,
   order_purchase_timestamp timestamp,
   order_approved_at timestamp,
   order_delivered_carrier_date timestamp,
   order_delivered_customer_date timestamp,
   order_estimated_delivery_date timestamp
);
```

```sql
delete from ecommerce_dev.gold.dim_date;

insert into ecommerce_dev.gold.dim_date(order_id,order_purchase_timestamp,order_approved_at,order_delivered_carrier_date,order_delivered_customer_date,order_estimated_delivery_date)
select order_id,order_purchase_timestamp,order_approved_at,order_delivered_carrier_date,order_delivered_customer_date,order_estimated_delivery_date
from ecommerce_dev.silver.orders_enhanced;

select * from ecommerce_dev.gold.dim_date;
```

```sql
delete from ecommerce_dev.gold.dim_geography;

insert into ecommerce_dev.gold.dim_geography(customer_id,customer_zip_code_prefix,customer_city,customer_state)
select customer_id,customer_zip_code_prefix,customer_city,customer_state from ecommerce_dev.silver.customers_enhanced;

select * from ecommerce_dev.gold.dim_geography;
```
