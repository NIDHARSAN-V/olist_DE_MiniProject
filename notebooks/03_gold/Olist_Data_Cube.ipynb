```sql
CREATE OR REPLACE TABLE ecommerce_dev.gold.olist_datacube AS
SELECT
  fo.order_id,
  fo.customer_id,
  fo.order_purchase_timestamp,
  fo.order_month,
  fo.order_year,
  fo.order_status,
  fo.is_delivered,
  fo.is_late_delivery,
  foi.order_item_id,
  foi.product_id,
  foi.seller_id,
  foi.total_item_value,
  fp.payment_type,
  fp.total_payment_value,
  fr.review_id,
  fr.review_score,
  fr.review_category,
  dc.customer_unique_id,
  dg.customer_city,
  dg.customer_state,
  dp.product_category_name,
  ds.seller_city AS seller_city,
  ds.seller_state AS seller_state,
  dd.order_purchase_timestamp AS purchase_ts,
  dd.order_delivered_customer_date,
  dd.order_estimated_delivery_date
FROM ecommerce_dev.gold.fact_order fo
LEFT JOIN ecommerce_dev.gold.fact_order_items foi ON fo.order_id = foi.order_id
LEFT JOIN ecommerce_dev.gold.fact_payment fp ON fo.order_id = fp.order_id
LEFT JOIN ecommerce_dev.gold.fact_review fr ON fo.order_id = fr.order_id AND foi.product_id = fr.product_id
LEFT JOIN ecommerce_dev.gold.dim_customer dc ON fo.customer_id = dc.customer_id
LEFT JOIN ecommerce_dev.gold.dim_geography dg ON fo.customer_id = dg.customer_id
LEFT JOIN ecommerce_dev.gold.dim_product dp ON foi.product_id = dp.product_id
LEFT JOIN ecommerce_dev.gold.dim_seller ds ON foi.seller_id = ds.seller_id
LEFT JOIN ecommerce_dev.gold.dim_date dd ON fo.order_id = dd.order_id
```
