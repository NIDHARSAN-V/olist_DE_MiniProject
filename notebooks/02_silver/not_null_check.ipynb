```python
from pyspark.sql.functions import col, when, isnull, sum as _sum
from pyspark.sql import Row

table_name = "ecommerce_dev.silver.customers_enhanced"
df = spark.table(table_name)
total = df.count()

null_counts = {c: df.where(col(c).isNull()).count() for c in df.columns}
null_cols = {k: v for k, v in null_counts.items() if v > 0}

if null_cols:
    result = spark.createDataFrame([Row(column_name=k, null_count=v, null_pct=round(v/total*100, 2)) for k, v in null_cols.items()])
    print(f"Table: {table_name} | Total rows: {total}")
    display(result)
else:
    print(f"✅ No null values found in {table_name} ({total} rows)")
```
